#### GPT OSS open models
Testing out common use cases for the gpt open models using the huggingface library

```python
import transformers
transformers.__version__
```

```python
import torch
from transformers import pipeline
```

```python
model_id = "openai/gpt-oss-20b"

pipe = pipeline(
    "text-generation",
    model=model_id,
    torch_dtype="auto",
    device_map="auto",
)

messages = [
    {"role": "user", "content": "Explain quantum mechanics clearly and concisely."},
]

outputs = pipe(
    messages,
    max_new_tokens=256,
)
print(outputs[0]["generated_text"][-1])
```
